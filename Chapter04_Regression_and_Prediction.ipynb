## 1. Chapter Summary

**Regression and prediction** are statistical approaches used to **describe relationships between variables and estimate outcomes from predictors,** enabling structured analysis and informed decision‑making in diverse contexts.

This chapter discusses linear regression, prediction intervals, factor variables, coefficient interpretation, regression diagnostics, and nonlinear modeling approaches such as polynomial and spline regression.

The concepts provide a foundation for building predictive models, interpreting variable relationships, evaluating assumptions, and identifying limitations that may influence regression‑based conclusions and analytical accuracy.

## 2. Learning Objectives

This chapter aims to provide an understanding of:

- Explaining simple linear regression with equations, residuals, and least squares.
- Applying multiple regression for predictors, performance, validation, and model selection.
- Using regression for prediction, avoiding extrapolation, and interpreting intervals.
- Representing factor variables with dummy variables for categories and levels.
- Interpreting regression with correlated predictors, multicollinearity, confounding, and interactions.
- Conducting diagnostics to detect outliers, influence, heteroskedasticity, non‑normality, and nonlinear patterns.
- Applying polynomial, spline, and additive models for nonlinear relationships.
