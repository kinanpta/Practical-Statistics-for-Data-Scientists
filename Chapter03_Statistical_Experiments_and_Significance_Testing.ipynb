## 1. Chapter Summary

**Statistical experiments and significance testing** provide structured methods for **evaluating differences, associations, and treatment effects while accounting for random variation in observed data.**

This chapter include A/B testing, hypothesis tests, resampling techniques, statistical significance, p‑values, t‑tests, multiple testing, degrees of freedom, ANOVA, chi‑square procedures, multi‑arm bandit strategies, and statistical power.

These concepts provide essential knowledge for evaluating experimental outcomes, interpreting statistical evidence, and understanding limitations inherent in conclusions derived from hypothesis testing and comparative analysis.

## 2. Learning Objectives

This chapter aims to provide an understanding of:

- Explaining principles of A/B testing and control groups.
- Formulating null and alternative hypotheses in one‑way and two‑way tests.
- Applying resampling and permutation for group differences.
- Interpreting statistical significance, p‑values, and error types in decision‑making.
- Using t‑tests for comparing sample means and assessing evidence.
- Addressing multiple testing and its implications.
- Understanding degrees of freedom in estimation and hypothesis testing.
- Conducting analysis of variance (ANOVA) with F‑statistics and two‑way designs.
- Applying chi‑square and Fisher’s exact for categorical data.
- Utilizing multi‑arm bandit algorithms to balance exploration and exploitation.
- Evaluating statistical power, sample size, and detecting meaningful effects.
