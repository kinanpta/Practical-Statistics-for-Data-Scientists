## 1. Chapter Summary

**Unsupervised learning** is an analytical approach that **identifies patterns, structures, and relationships in datasets without requiring a labeled response variable.**

This chapter introduces principal component analysis (PCA), K-means, hierarchical clustering, model-based clustering, and techniques for handling scaling issues and categorical variables in mixed‑data contexts.

This chapter’s concepts provide a basis for reducing dimensionality, identifying clusters, exploring complex structures, and interpreting similarities among observations, supporting comprehensive analysis across varied data types and contexts.

## 2. Learning Objectives

This chapter aims to provide an understanding of:

- Applying principal component analysis for computing, interpreting, and correspondence analysis.
- Using K‑means clustering for assignment, interpretation, and K selection.
- Explaining hierarchical clustering with dendrograms, agglomerative methods, and dissimilarity.
- Implementing model‑based clustering with multivariate normals, mixtures, and cluster choice.
- Addressing scaling and categorical variables with Gower’s distance and mixed‑data issues.
